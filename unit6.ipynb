wap such that your should prompt for a file name and display first fifty lines at a time then pausing it to ask the user to continue or stop (pager program)

```python
file=input('enter file name')
with open (file,'r') as f:
    for i in range (50):
        print(f.readline(),end=' ')
    while(True):
        inp=input('s for stop otherwise continue')
        if(inp==('s'or'S')):
            break
        else:
            for i in range (50):
                print(f.readline(),end=' ')
```

wap to copy a code of file to another but it should not have any comments in it

```python
f1=open('p1.txt','r')
f2=open('p4.txt','w')
l=f1.readlines()
for i in l:
    if('#' in i):
        x=i.split('#')
        f2.write(x[0]+'\n')
    else:
        f2.write(i)
f2.close()
```

wap to compare a file and print from where content changes

```python
f1=open('p1.txt','r')
f2=open('p5.txt','r')
a=f1.readlines()
b=f2.readlines()
for index,line in enumerate(a):
    if line != b[index]:
        for ind,char in enumerate(line):
            if char != b[index][ind]:
                print(char)
                print('row number:',index+1,'column number:',ind+1)
                break
        break
else:
    print('Same')
```
